
# CryptoBuddy: Your Eco-Friendly Crypto Advisor
A rule-based chatbot that recommends cryptos based on profitability and sustainability.
> Add blockquote



```python
# Predefined Crypto Database
crypto_db = {
    "Bitcoin": {
        "price_trend": "rising",
        "market_cap": "high",
        "energy_use": "high",
        "sustainability_score": 3/10
    },
    "Ethereum": {
        "price_trend": "stable",
        "market_cap": "high",
        "energy_use": "medium",
        "sustainability_score": 6/10
    },
    "Cardano": {
        "price_trend": "rising",
        "market_cap": "medium",
        "energy_use": "low",
        "sustainability_score": 8/10
    },
    "Solana": {  # Bonus entry for testing
        "price_trend": "rising",
        "market_cap": "medium",
        "energy_use": "low",
        "sustainability_score": 7/10
    }
}

print("Data loaded! Example:", list(crypto_db.keys()))
```

```python
# Main chatbot loop
def run_chatbot(crypto_db):
    print("CryptoBuddy: Hey there, crypto explorer! 🚀 I'm here to help with profitability and sustainability advice.")
    print("CryptoBuddy: Type 'bye' to exit. Remember, this is not financial advice—crypto is volatile. Do your own research! ⚠️")

    import pandas as pd  # Import pandas here

    # Rule-based recommendation functions
    def get_profitable_recommendation():
        profitable = [coin for coin in crypto_db
                      if crypto_db[coin]["price_trend"] == "rising" and
                      crypto_db[coin]["market_cap"] == "high"]
        return profitable[0] if profitable else "None—wait for the dip! 📉"

    def get_sustainable_recommendation():
        sustainable = [coin for coin in crypto_db
                       if crypto_db[coin]["energy_use"] == "low" and
                       crypto_db[coin]["sustainability_score"] > 7/10]
        return sustainable[0] if sustainable else "Look for greener options! 🌍"


    while True:
        user_input = input("You: ").lower().strip()
        if user_input == 'bye':
            print("CryptoBuddy: Safe travels in the crypto world! 🌟")
            break
        # Enhanced chatbot with logic
        if "trending" in user_input or "rising" in user_input or "growth" in user_input:
            # Find rising trends
            rising_coins = [coin for coin in crypto_db if crypto_db[coin]["price_trend"] == "rising"]
            if rising_coins:
                recommend = rising_coins[0]  # Pick first for simplicity
                print(f"CryptoBuddy: For growth, check out {recommend}! It's rising 🚀 with {crypto_db[recommend]['market_cap']} market cap.")
            else:
                print("CryptoBuddy: No rising coins right now—market's chill. 😎")

        elif  "sustainable" in user_input or "eco" in user_input or "green" in user_input:
            # Find highest sustainability
            recommend = max(crypto_db, key=lambda x: crypto_db[x]["sustainability_score"])
            print(f"CryptoBuddy: Go green with {recommend}! 🌱 Score: {crypto_db[recommend]['sustainability_score']*10:.0f}/10, low energy use.")

        elif "buy" in user_input or "recommend" in user_input:
            # In the elif "buy" block:
            profit_pick = get_profitable_recommendation()
            sustain_pick = get_sustainable_recommendation()
            print(f"CryptoBuddy: Profit-focused: {profit_pick}. Sustainable: {sustain_pick}. Balance both! ⚖️")
            # Balanced advice (profit + sustain)
            balanced = [coin for coin in crypto_db
                        if crypto_db[coin]["price_trend"] == "rising" and
                        crypto_db[coin]["sustainability_score"] > 0.7]
            if balanced:
                recommend = balanced[0]
                print(f"CryptoBuddy: For long-term win, {recommend} rocks! Rising trend + eco-score {crypto_db[recommend]['sustainability_score']*10:.0f}/10. 💚")
            else:
                print("CryptoBuddy: Tough market—stick to research! 📚")

        elif "help" in user_input:
            print("CryptoBuddy: Ask about 'trending', 'sustainable', or 'recommendations'! 😊")

        else:
            print("CryptoBuddy: Hmm, tell me more—like 'What's sustainable?' or 'Trending coins?' 🤔")

# Run the bot
run_chatbot(crypto_db)
```
